```python
df1=pd.read_csv("Departments (1).csv")
df2=pd.read_csv("Employees (1).csv")
df3=pd.read_csv("Jobs (1).csv")
df4=pd.read_csv("JobsHistory (1).csv")
df5=pd.read_csv("Locations (1).csv")
df1.to_sql("dep",con,if_exists="replace",index=False)
df2.to_sql("emp",con,if_exists="replace",index=False)
df3.to_sql("job",con,if_exists="replace",index=False)
df4.to_sql("his",con,if_exists="replace",index=False)
df5.to_sql("loc",con,if_exists="replace",index=False)
cur=con.cursor()
cur.execute("select name from sqlite_master where type='table'")
print(cur.fetchall())
```
